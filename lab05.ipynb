## Plot 1

**Scenario**: A researcher wants to show that Gentoo penguins are significantly heavier than other species, but their plot is visually overwhelming and exaggerates the differences.

```python
penguins = sns.load_dataset("penguins")

ax = sns.barplot(data=penguins, x="species", y="body_mass_g", palette="hsv")
ax.set_ylim(0, 5500) 
plt.title("Penguin Mass by Species")
plt.show()
```

Issues with the plot: Gentoo looks like have triple mass compared to other penguin.


How you changed the plot and why: Changed the y-axis of graph to start from 0 because now it is not overwhelming and easier to understand. In this new graph, the Getoo penguin does not look like have triple mass, but with the graph that start with 0, it's more "not steep." 


## Plot 2

**Scenario**: A student plotted the carat size versus price for 54,000 diamonds to see where the bulk of the market sits, but the result is just a solid block of ink.

```python
diamonds = sns.load_dataset("diamonds")

sns.scatterplot(data=diamonds, x="carat", y="price", color="blue", s=5)
plt.title("Diamond Price vs. Carat")
plt.show()
```

Issues with the plot: We can not see anything and just white.


How you changed the plot and why: Made s=100 to s=1. This change made the graph visible and now we can see the obvious line at around carat = 1, 1.5, 2. When s=100, the graph was not so visible because probably, the each point has white line of a point overlaps and became just one big white... something.


## Plot 3

The aviation board wants to see passenger growth month-over-month for each year in the 1950s, but the lines are indistinguishable.

```python
flights = sns.load_dataset("flights")

sns.lineplot(data=flights, x="month", y="passengers", units="year", estimator=None, hue="year", palette="hsv")
plt.title("Passengers over Time")
plt.show()
```

Issues with the plot: We can not distinguish between the line and which year is which.


Changed color="black" to hue="year", palette="hsv" now so that the color change for each year and label year from 1950 to 1960 so that the line is distinguishable.

## Plot 4

**Scenario**: A restaurant manager receives this chart showing the relationship between total bills and tips, split by smoking preference. 





```python
tips = sns.load_dataset("tips")

sns.lmplot(data=tips, x="total_bill", y="tip", hue="smoker", scatter_kws={"s": 5}, line_kws={"linewidth": 1})
plt.show()
```

Issues with the plot: It is hard to see and where the points are. 


How you changed the plot and why: Made each dots smaller and width of line to 1 because it was hard to see so that now we can see the intersection which can be used for descision of conclusion to be made easier and efficiently.
